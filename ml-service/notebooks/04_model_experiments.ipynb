# 04 — Model Experiments

**Goal:** Train, tune, and evaluate pricing models. Compare baselines vs LightGBM quantile.

**Input:** `data/processed/features.parquet`  
**Output:** Trained models in `models/v*/`, evaluation metrics, SHAP analysis

## Experiment Plan
1. Train/test split (80/20, stratified by brand)
2. **Baseline models:** Linear Regression, Random Forest → establish floor metrics
3. **LightGBM median (α=0.5):** Tune with Optuna (100 trials)
4. **LightGBM quantile models:** α=0.1 (floor), α=0.5 (fair), α=0.9 (ceiling)
5. **Log-price experiment:** Train on log(price), compare MAPE
6. Cross-validation (5-fold) on best model
7. **SHAP analysis:** Feature importance, waterfall plots, dependence plots
8. **Evaluation:** MAPE, MAE, R², coverage rate (% of actuals within [α=0.1, α=0.9])
9. **Segmented analysis:** Metrics per brand, per price bracket
10. Save best models + registry update
