# 03 — Feature Engineering

**Goal:** Transform cleaned data into a training-ready feature matrix.

**Input:** `data/processed/cleaned.parquet` + `data/lookups/car_specs_lookup.csv` + `data/lookups/title_parsed.csv`  
**Output:** `data/processed/features.parquet` (ready for model training)

## Steps
1. Load cleaned data + lookup CSVs
2. Join with `title_parsed.csv` → get `brand`, `model` columns
3. Join with `car_specs_lookup.csv` → get `engine_cc`, `hp`, `body_type`, `new_car_price_egp`
4. Engineer derived features:
   - `car_age` = current_year - year
   - `depreciation_ratio` = price / new_car_price_egp
   - `mileage_per_year` = mileage / car_age
   - `hp_per_cc` = hp / engine_cc
   - `is_high_mileage` = mileage > 150,000
5. Handle remaining NaNs (median imputation for numeric, mode for categorical)
6. Encode categoricals for LightGBM (label encode or use native categorical)
7. Final correlation check + save
